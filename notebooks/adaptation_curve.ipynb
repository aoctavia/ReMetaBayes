# 📘 Title: Adaptation Curve Analysis for ReMetaBayes
# Visualize agent recovery speed after distribution shifts


```python
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# -------------------------------
# Load data (example structure)
# -------------------------------

```python
RESULT_PATH = "../results/gridworld/adaptation_metrics.json"

if not os.path.exists(RESULT_PATH):
    # Dummy synthetic adaptation example
    np.random.seed(42)
    data = {
        "shift_id": np.arange(1, 6),
        "pre_shift_reward": np.random.uniform(0.2, 0.6, 5).tolist(),
        "post_shift_reward": np.random.uniform(0.1, 0.5, 5).tolist(),
        "recovery_steps": np.random.randint(20, 100, 5).tolist()
    }
    with open(RESULT_PATH, "w") as f:
        json.dump(data, f)

with open(RESULT_PATH, "r") as f:
    data = json.load(f)

df = pd.DataFrame(data)
```

# -------------------------------
# Compute Adaptation Index
# -------------------------------

```python
df["adaptation_index"] = (df["post_shift_reward"] - df["pre_shift_reward"]) / df["recovery_steps"]

print("Adaptation Metrics:\n", df.round(3))
```

# -------------------------------
# 1️⃣ Recovery Curve
# -------------------------------

```python
plt.figure(figsize=(8,4))
sns.barplot(x="shift_id", y="recovery_steps", data=df, palette="Blues_d")
plt.title("Recovery Steps after Distribution Shift", fontsize=14)
plt.xlabel("Shift Event")
plt.ylabel("Steps to Recover")
plt.tight_layout()
plt.show()
```

# -------------------------------
# 2️⃣ Adaptation Index
# -------------------------------

```python
plt.figure(figsize=(8,4))
sns.lineplot(x="shift_id", y="adaptation_index", data=df, marker="o", color="tab:green", linewidth=2)
plt.title("Adaptation Index per Shift", fontsize=14)
plt.xlabel("Shift Event")
plt.ylabel("Adaptation Index (Δreward / steps)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
