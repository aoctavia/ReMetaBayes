# 📘 Title: ReMetaBayes Visualization
# This notebook visualizes experiment results from ReMetaBayes (GridWorld or Bandit).


```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Optional: W&B if online visualization is needed

```python
try:
    import wandb
    wandb.login()
except Exception:
    print("W&B not configured. Local visualization only.")

# -------------------------------
# Load experiment results
# -------------------------------
RESULT_PATH = "../results/gridworld/metrics.json"

if os.path.exists(RESULT_PATH):
    with open(RESULT_PATH, "r") as f:
        data = json.load(f)
else:
    # Dummy example if no file yet
    data = {
        "episode": list(range(1000)),
        "reward": np.random.normal(0, 1, 1000).tolist(),
        "uncertainty": np.random.rand(1000).tolist(),
        "kl_divergence": np.abs(np.random.normal(0, 0.05, 1000)).tolist()
    }

df = pd.DataFrame(data)
```

# -------------------------------
# 1️⃣ Reward Curve
# -------------------------------

```python
plt.figure(figsize=(8,4))
sns.lineplot(x="episode", y="reward", data=df, color="tab:blue", linewidth=1.5)
plt.title("Average Episode Reward over Time", fontsize=14)
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# -------------------------------
# 2️⃣ Uncertainty Evolution
# -------------------------------

```python
plt.figure(figsize=(8,4))
sns.lineplot(x="episode", y="uncertainty", data=df, color="tab:orange", linewidth=1.5)
plt.title("Uncertainty Evolution (Posterior Variance)", fontsize=14)
plt.xlabel("Episode")
plt.ylabel("Uncertainty")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# -------------------------------
# 3️⃣ Posterior Drift (KL divergence)
# -------------------------------

```python
plt.figure(figsize=(8,4))
sns.lineplot(x="episode", y="kl_divergence", data=df, color="tab:red", linewidth=1.5)
plt.title("Posterior Drift across Non-stationary Shifts", fontsize=14)
plt.xlabel("Episode")
plt.ylabel("KL(q(θ_t)||q(θ_{t-1}))")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# -------------------------------
# 4️⃣ Summary Heatmap (optional TheoRL-Bench view)
# -------------------------------

```python
if "reward" in df and "uncertainty" in df:
    corr = df[["reward", "uncertainty", "kl_divergence"]].corr()
    plt.figure(figsize=(4,3))
    sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
    plt.title("Metric Correlation Matrix")
    plt.tight_layout()
    plt.show()
```
